Implement game assignments, venue locations, and movement costs using Gurobi. The demonstration uses a simplified four-team tournament with one game per directed matchup and no rest days. See the [mathematical formulation](../docs/mathematical_formulation.md).

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
```

```python
from src.example_data import four_team_instance
from src.integer_programming import solve
from src.validation import validate, travel_cost
costs, demand = four_team_instance()
try:
    result = solve(costs, k=1, time_limit=60)
except (ImportError, RuntimeError) as exc:
    print('Gurobi unavailable:', exc)
    result = None
if result is not None:
    print({k:v for k,v in result.items() if k!='schedule'})
    if result['schedule'] is not None:
        print(validate(result['schedule'],demand))
        print('Independent cost:',travel_cost(result['schedule'],costs))
```
