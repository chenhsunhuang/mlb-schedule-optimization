Analyze the nine Gurobi experiments reported in the original course project to examine how computational performance changes with problem size.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
```

```python
import pandas as pd
from IPython.display import display, Image

results = pd.read_csv(ROOT / 'results/reported_gurobi_results.csv')
display(results)
```

## Runtime by Problem Size

The plot distinguishes completed runs, time-limited runs, and resource failures. No runtime is assigned to resource-failure cases.

```python
display(Image(filename=str(ROOT / 'figures/gurobi_runtime.png')))
summary = results.groupby('teams', sort=True).agg(
    scenarios=('scenario', 'count'),
    reported_runtimes=('gurobi_seconds', 'count'),
    reported_costs=('reported_gurobi_cost_usd', 'count'),
)
display(summary)
```

The four-team runs completed, while larger instances reached the solver's time limit or encountered computational resource constraints, highlighting the scalability challenges of integer programming in large-scale scheduling problems.
